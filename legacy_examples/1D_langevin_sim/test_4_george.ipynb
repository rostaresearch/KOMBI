```python
def get_working_MM(M):
    zero_rows = np.where(~M.any(axis=1))[0]
    zero_cols = np.where(~M.any(axis=0))[0]
    print(zero_cols, zero_rows)
    print(np.union1d(zero_rows, zero_cols))

    keep_indices = np.setdiff1d(range(M.shape[0]), np.union1d(zero_rows, zero_cols))
    print(keep_indices)
    M_work = M[np.ix_(keep_indices, keep_indices)]
    return M_work, keep_indices

def get_working_MM_intersect(M):
    zero_rows = np.where(~M.any(axis=1))[0]
    zero_cols = np.where(~M.any(axis=0))[0]

    keep_indices = np.intersect1d(range(M.shape[0]), np.union1d(zero_rows, zero_cols))
    M_work = M[np.ix_(keep_indices, keep_indices)]
    return M_work, keep_indices

def get_working_MM_v2(M):
    # Identify non-zero rows and columns
    non_zero_rows = np.where(M.any(axis=1))[0]
    non_zero_cols = np.where(M.any(axis=0))[0]
    
    # Filter the matrix to keep only non-zero rows and columns
    M_work = M[np.ix_(non_zero_rows, non_zero_cols)]
    
    return M_work, non_zero_rows, non_zero_cols
```
